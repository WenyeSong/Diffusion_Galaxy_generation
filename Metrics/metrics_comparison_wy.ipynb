## Table 1 Reproduction — paper format

```python

# score = mean(generated) / mean(real test), closer to 1 = better
#   A) Paper Gen / Paper Real  → from their own CSVs (may differ from Table 1 — see note)
#   B) Reproduce / Paper Real  → our generated vs paper's real baseline
#   C) Reproduce / Paper Gen   → how close our generation is to theirs

import pandas as pd
import numpy as np

# Paper Table 1 reported values (σ=0.1, their best model)
PAPER_TABLE1 = {
    'Ellipticity':                          0.98,
    'Semi-major Axis':                      0.96,
    'Ellipticity Proxy (Sérsic Index in paper)': 0.93,
    'Isophotal Area':                       0.90,
}

# Column names in the CSVs
CSV_COL = {
    'Ellipticity':                          'Ellipticity',
    'Semi-major Axis':                      'Semi-major Axis',
    'Ellipticity Proxy (Sérsic Index in paper)': 'Sersic Index',
    'Isophotal Area':                       'Isophotal Area',
}

METRICS = list(PAPER_TABLE1.keys())

rows = []
for m in METRICS:
    col = CSV_COL[m]
    r  = paper_real[col].dropna().mean()
    pg = paper_gen[col].dropna().mean()
    rg = repro_gen[col].dropna().mean()

    ratio_A = pg / r
    ratio_B = rg / r
    ratio_C = rg / pg
    reported = PAPER_TABLE1[m]
    match = '✓' if abs(ratio_A - reported) < 0.03 else '✗'

    rows.append({
        'Metric':                        m,
        'Paper Real mean':               round(r,  4),
        'Paper Gen mean':                round(pg, 4),
        'Reproduce mean':                round(rg, 4),
        'A: PaperGen/PaperReal':         round(ratio_A, 4),
        'Paper Table1 (σ=0.1)':          reported,
        'Match?':                        match,
        'B: Repro/PaperReal':            round(ratio_B, 4),
        'C: Repro/PaperGen':             round(ratio_C, 4),
    })

df = pd.DataFrame(rows)

print("=" * 95)
print("Table 1 Reproduction")
print("  score = mean(generated) / mean(real test)   |   closer to 1.0 = better")
print("=" * 95)
print(f"{'Metric':<42} {'A:PaperGen/Real':>16} {'Paper Table1':>13} {'Match':>6}  "
      f"{'B:Repro/Real':>13} {'C:Repro/PaperGen':>17}")
print("-" * 95)
for r in rows:
    print(f"{r['Metric']:<42} {r['A: PaperGen/PaperReal']:>16.4f} "
          f"{r['Paper Table1 (σ=0.1)']:>13} {r['Match?']:>6}  "
          f"{r['B: Repro/PaperReal']:>13.4f} {r['C: Repro/PaperGen']:>17.4f}")

print()
print("NOTE: A ≠ Paper Table1 is expected — likely causes:")
print("  1. AndrewMetrics.csv has 39k rows (not 10k), may be from a different model/run")
print("  2. Paper used 10k sampled test images as denominator, not all 40,914")
print("  3. AndrewMetrics might use different σ or filtering than paper's best model")
print()
print("B and C are our reproduction results using the paper's CSVs as baseline.")

df.to_csv('./table1_reproduction.csv', index=False)
print("Saved: table1_reproduction.csv")
```

```python
## Table 1 — Z-matched ratio (approximating paper's method)
# Paper: sample 10k z from test set → generate images → ratio = gen / matched_real
# We:    for each generated image, find nearest-z test image → ratio = gen / matched_real
#
# Both numerator and denominator now have the SAME redshift distribution → fair comparison

import pandas as pd
import numpy as np

repro_gen_m = pd.read_csv("./generated_images_metrics_wy.csv")   # has Redshift
our_test_m  = pd.read_csv("./testing_images_metrics_wy.csv")     # has Redshift

print(f"Generated: {len(repro_gen_m)} images  z range [{repro_gen_m['Redshift'].min():.2f}, {repro_gen_m['Redshift'].max():.2f}]")
print(f"Test set:  {len(our_test_m)} images  z range [{our_test_m['Redshift'].min():.2f}, {our_test_m['Redshift'].max():.2f}]")

# For each generated image, find the nearest-z test image
gen_z  = repro_gen_m['Redshift'].values
test_z = our_test_m['Redshift'].values

print("\nMatching generated images to nearest-z test images...")
matched_idx = np.array([np.argmin(np.abs(test_z - z)) for z in gen_z])
matched_test = our_test_m.iloc[matched_idx].reset_index(drop=True)

# Check matching quality
dz = np.abs(gen_z - test_z[matched_idx])
print(f"Median |Δz| between gen and matched test: {np.median(dz):.4f}")
print(f"Max    |Δz|: {dz.max():.4f}")

# Compute ratios
METRICS = ['Ellipticity', 'Semi-major Axis', 'Sersic Index', 'Isophotal Area']
PAPER_TABLE1 = {'Ellipticity': 0.98, 'Semi-major Axis': 0.96,
                'Sersic Index': 0.93, 'Isophotal Area': 0.90}

print("\n" + "="*80)
print("Z-matched Table 1  (ratio = mean(gen) / mean(z-matched real test))")
print("="*80)
print(f"{'Metric':<30} {'Gen mean':>10} {'Matched Real':>13} {'Ratio':>8} {'Paper':>7} {'Δ':>7}")
print("-"*80)

rows = []
for m in METRICS:
    gen_mean  = repro_gen_m[m].dropna().mean()
    real_mean = matched_test[m].dropna().mean()
    ratio     = gen_mean / real_mean
    paper     = PAPER_TABLE1.get(m, float('nan'))
    delta     = ratio - paper
    print(f"{m:<30} {gen_mean:>10.4f} {real_mean:>13.4f} {ratio:>8.4f} {paper:>7} {delta:>+7.4f}")
    rows.append({'Metric': m, 'Gen mean': round(gen_mean,4),
                 'Matched Real mean': round(real_mean,4),
                 'Z-matched ratio': round(ratio,4),
                 'Paper Table1': paper, 'Delta': round(delta,4)})

print()
print("Δ = our ratio − paper value  (closer to 0 = better reproduction)")
pd.DataFrame(rows).to_csv('./table1_zmatched.csv', index=False)
print("Saved: table1_zmatched.csv")
```

```python
## Why A gives 1.4: z-distribution mismatch proof
# If z distributions differ, means are not comparable.
# Check: per-bin ratios should be closer to 1 than global ratio.

import pandas as pd
import numpy as np

repro_gen_m = pd.read_csv("./generated_images_metrics_wy.csv")
our_test_m  = pd.read_csv("./testing_images_metrics_wy.csv")
paper_real  = pd.read_csv("./testing_metrics.csv")

# z distributions
print("=== z distribution comparison ===")
print(f"testing_metrics.csv (paper_real): mean z = {paper_real['Sequential Index'].head(0).values}  (no z column — can't check)")
print(f"generated_images_wy (repro_gen):  mean z = {repro_gen_m['Redshift'].mean():.3f}")
print(f"testing_images_wy   (our_test):   mean z = {our_test_m['Redshift'].mean():.3f}")

print("\n=== Ellipticity per z-bin: gen vs real (should be ~1 within each bin) ===")
bins = [0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 4.0]
print(f"{'z bin':<12} {'Gen mean':>10} {'Real mean':>10} {'Ratio':>8} {'N_gen':>7} {'N_real':>7}")
print("-"*58)
for lo, hi in zip(bins[:-1], bins[1:]):
    g = repro_gen_m[(repro_gen_m['Redshift'] >= lo) & (repro_gen_m['Redshift'] < hi)]['Ellipticity'].dropna()
    r = our_test_m[(our_test_m['Redshift'] >= lo) & (our_test_m['Redshift'] < hi)]['Ellipticity'].dropna()
    if len(g) > 0 and len(r) > 0:
        ratio = g.mean() / r.mean()
        print(f"z={lo:.1f}-{hi:.1f}    {g.mean():>10.4f} {r.mean():>10.4f} {ratio:>8.4f} {len(g):>7} {len(r):>7}")

print("\nGlobal ratio (wrong): gen.mean / real.mean =",
      round(repro_gen_m['Ellipticity'].mean() / our_test_m['Ellipticity'].mean(), 4))
print("→ per-bin ratios above should all be closer to 1 than the global 1.4")
```

```python
## Table 1 — Sample 10k z from test set (paper-equivalent method)
# 1. Sample 10,000 z values from real test set distribution
# 2. Match BOTH gen and real to those same z values
# 3. ratio = mean(gen) / mean(real)  — same z distribution → fair comparison
# 4. score = min(ratio, 1/ratio)    — symmetric closeness to 1 (like paper's format)
# Also includes: Andrew/PaperReal (global, no z col in AndrewMetrics)

import pandas as pd
import numpy as np

repro_gen_m = pd.read_csv("./generated_images_metrics_wy.csv")
our_test_m  = pd.read_csv("./testing_images_metrics_wy.csv")
paper_real  = pd.read_csv("./testing_metrics.csv")
andrew      = pd.read_csv("./AndrewMetrics.csv")

np.random.seed(42)
N = 10000

# ── z-matched sampling ────────────────────────────────────────────────────────
sample_z = np.random.choice(our_test_m['Redshift'].values, size=N, replace=False)

test_z = our_test_m['Redshift'].values
real_idx = np.array([np.argmin(np.abs(test_z - z)) for z in sample_z])
matched_real = our_test_m.iloc[real_idx].reset_index(drop=True)

gen_z = repro_gen_m['Redshift'].values
gen_idx = np.array([np.argmin(np.abs(gen_z - z)) for z in sample_z])
matched_gen = repro_gen_m.iloc[gen_idx].reset_index(drop=True)

dz_real = np.abs(sample_z - test_z[real_idx])
dz_gen  = np.abs(sample_z - gen_z[gen_idx])
print(f"Sampled z: mean={sample_z.mean():.3f}  std={sample_z.std():.3f}")
print(f"Real match quality: median|Δz|={np.median(dz_real):.4f}")
print(f"Gen  match quality: median|Δz|={np.median(dz_gen):.4f}")

# ── metric name mapping ───────────────────────────────────────────────────────
METRICS = [
    ('Ellipticity Proxy (Sérsic Index in paper)', 'Sersic Index', 'Sersic Index'),
    ('Ellipticity',                               'Ellipticity',  'Ellipticity'),
    ('Semi-major Axis',                           'Semi-major Axis', 'Semi-major Axis'),
    ('Isophotal Area',                            'Isophotal Area',  'Isophotal Area'),
]
PAPER = {
    'Ellipticity Proxy (Sérsic Index in paper)': 0.93,
    'Ellipticity':     0.98,
    'Semi-major Axis': 0.96,
    'Isophotal Area':  0.90,
}

# ── print table ───────────────────────────────────────────────────────────────
print("\n" + "="*105)
print(f"Table 1 comparison  (N={N:,} z-sampled; score = min(ratio, 1/ratio))")
print("="*105)
print(f"{'Metric':<42} {'Repro score':>12} {'Andrew/Real':>12} {'Paper':>7} {'Δ repro':>9} {'Δ andrew':>10}")
print("-"*105)

rows = []
for label, gen_col, andrew_col in METRICS:
    # A) Reproduce z-matched score
    gm = matched_gen[gen_col].dropna().mean()
    rm = matched_real[gen_col].dropna().mean()
    ratio_repro = gm / rm
    score_repro = min(ratio_repro, 1/ratio_repro)

    # B) Andrew / PaperReal (global — no z col in AndrewMetrics)
    am = andrew[andrew_col].dropna().mean()
    pm = paper_real[andrew_col].dropna().mean()
    ratio_andrew = am / pm
    score_andrew = min(ratio_andrew, 1/ratio_andrew)

    paper = PAPER[label]
    print(f"{label:<42} {score_repro:>12.4f} {score_andrew:>12.4f} {paper:>7} "
          f"{score_repro-paper:>+9.4f} {score_andrew-paper:>+10.4f}")
    rows.append({'Metric': label,
                 'Repro score (z-matched)': round(score_repro, 4),
                 'Andrew/PaperReal score':  round(score_andrew, 4),
                 'Paper Table1':            paper,
                 'Δ Repro':                 round(score_repro - paper, 4),
                 'Δ Andrew':                round(score_andrew - paper, 4)})

print()
print("Repro   = our generated images vs z-matched real test (our SEP metrics)")
print("Andrew  = AndrewMetrics.csv vs testing_metrics.csv (global, no z matching — expected to differ)")
print("Δ: + means our score is higher than paper, − means lower; closer to 0 = better")

pd.DataFrame(rows).to_csv('./table1_zsampled.csv', index=False)
print("Saved: table1_zsampled.csv")
```

```python
## Final Summary Table
# Repro score  = min(gen/real, real/gen)  z-matched, OUR SEP test
# Paper score  = min(gen/real, real/gen)  global, HSC test (no z-match)
# Paper stated = Table 1 in Lizarraga et al. 2024
# Repro/Paper  = mean(our gen) / mean(AndrewMetrics)  — direct mean comparison

import pandas as pd
import numpy as np

repro_gen_m = pd.read_csv("./generated_images_metrics_wy.csv")
our_test_m  = pd.read_csv("./testing_images_metrics_wy.csv")
paper_real  = pd.read_csv("./testing_metrics.csv")
andrew      = pd.read_csv("./AndrewMetrics.csv")

np.random.seed(42)
N = 10000

sample_z = np.random.choice(our_test_m['Redshift'].values, size=N, replace=False)
test_z   = our_test_m['Redshift'].values
gen_z    = repro_gen_m['Redshift'].values
matched_real = our_test_m.iloc[
    np.array([np.argmin(np.abs(test_z - z)) for z in sample_z])].reset_index(drop=True)
matched_gen  = repro_gen_m.iloc[
    np.array([np.argmin(np.abs(gen_z - z)) for z in sample_z])].reset_index(drop=True)

METRICS = [
    ('Ellipticity Proxy (Sersic in paper)', 'Sersic Index',    0.93),
    ('Ellipticity',                         'Ellipticity',     0.98),
    ('Semi-major Axis',                     'Semi-major Axis', 0.96),
    ('Isophotal Area',                      'Isophotal Area',  0.90),
]

rows = []
for label, col, paper_stated in METRICS:
    # Repro score: z-matched, our SEP test
    gm = matched_gen[col].dropna().mean()
    rm = matched_real[col].dropna().mean()
    score_repro = min(gm/rm, rm/gm)

    # Paper score: global, HSC test
    am = andrew[col].dropna().mean()
    pm = paper_real[col].dropna().mean()
    score_paper = min(am/pm, pm/am)

    # Repro/Paper: direct mean ratio (our gen mean / andrew gen mean)
    full_gm = repro_gen_m[col].dropna().mean()   # all our gen images
    repro_over_paper = full_gm / am

    rows.append({
        'Metric':        label,
        'Repro score':   round(score_repro, 4),
        'Paper score':   round(score_paper, 4),
        'Paper stated':  paper_stated,
        'Repro/Paper':   round(repro_over_paper, 4),
    })

df = pd.DataFrame(rows)

print("Test set used:")
print("  Repro score  → our SEP test (testing_images_metrics_wy.csv), z-matched")
print("  Paper score  → HSC catalog  (testing_metrics.csv), global (no z-match)")
print("  Repro/Paper  → mean(our gen) / mean(AndrewMetrics)  [direct mean ratio]")
print()

header = f"{'Metric':<40} {'Repro score':>12} {'Paper score':>12} {'Paper stated':>13} {'Repro/Paper':>12}"
sep = "=" * len(header)
print(sep)
print(header)
print(sep)
for r in rows:
    print(f"{r['Metric']:<40} {r['Repro score']:>12.4f} {r['Paper score']:>12.4f} "
          f"{r['Paper stated']:>13.2f} {r['Repro/Paper']:>12.4f}")
print()
print("Repro/Paper: 1.0 = identical mean to paper's gen; >1 = our gen has higher mean")

df.to_csv('./table1_final.csv', index=False)
print("Saved: table1_final.csv")
display(df)
```

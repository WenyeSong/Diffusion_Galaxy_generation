# Mean Metrics Comparison
Compare means across: Paper Real, Paper Generated, and Reproduced Generated.

```python
from IPython.display import display
import pandas as pd

# Load all three CSVs
paper_real = pd.read_csv("./testing_metrics.csv")
paper_gen  = pd.read_csv("./AndrewMetrics.csv")
repro_gen  = pd.read_csv("./generated_images_metrics_wy.csv")

print(f"Paper Real (testing_metrics.csv):     {len(paper_real)} rows")
print(f"Paper Generated (AndrewMetrics.csv):  {len(paper_gen)} rows")
print(f"Reproduced Generated (_wy.csv):       {len(repro_gen)} rows")
```

## Paper Real — testing_metrics.csv

```python
display(paper_real.head())
paper_real_avg = paper_real.drop(columns=["Sequential Index", "Object ID"], errors="ignore").mean(numeric_only=True)
print("Averages (Paper Real):")
for col, avg in paper_real_avg.items():
    print(f"  {col}: {avg:.4f}")
```

## Paper Generated — AndrewMetrics.csv

```python
display(paper_gen.head())
paper_gen_avg = paper_gen.drop(columns=["Image Index"], errors="ignore").mean(numeric_only=True)
print("Averages (Paper Generated):")
for col, avg in paper_gen_avg.items():
    print(f"  {col}: {avg:.4f}")
```

## Reproduced Generated — generated_images_metrics_wy.csv

```python
display(repro_gen.head())
repro_gen_avg = repro_gen.drop(columns=["Image File", "Image Index", "Redshift"], errors="ignore").mean(numeric_only=True)
print("Averages (Reproduced Generated):")
for col, avg in repro_gen_avg.items():
    print(f"  {col}: {avg:.4f}")
```

## Summary Comparison Table

```python
metrics = ["Semi-major Axis", "Ellipticity", "Sersic Index", "Isophotal Area"]

rows = []
for m in metrics:
    r  = paper_real[m].dropna().mean()
    pg = paper_gen[m].dropna().mean()
    rg = repro_gen[m].dropna().mean()
    rows.append({
        "Metric": m,
        "Paper Real mean": round(r, 4),
        "Paper Gen mean":  round(pg, 4),
        "Reproduce mean":  round(rg, 4),
        "Paper Gen / Real": round(pg/r, 4),
        "Reproduce / Real": round(rg/r, 4),
        "Reproduce / Paper": round(rg/pg, 4),
    })

summary = pd.DataFrame(rows)
display(summary)
summary.to_csv("./summary_means.csv", index=False)
print("Saved: summary_means.csv")
```

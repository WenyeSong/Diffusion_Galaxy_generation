```python
import pandas as pd
df = pd.read_csv('/Users/wen/Desktop/project/lizarraga_2024-main/Metrics/testing_images_metrics_wy_fit_sersic.csv')

print("Ellipticity Proxy:")
print("  有:", df['Ellipticity Proxy'].notna().sum())
print("  NaN:", df['Ellipticity Proxy'].isna().sum())

print("\nSersic Index (fitted):")
print("  有:", df['Sersic Index (fitted)'].notna().sum())
print("  NaN:", df['Sersic Index (fitted)'].isna().sum())

print("\n两者都有:", (df['Ellipticity Proxy'].notna() & df['Sersic Index (fitted)'].notna()).sum())
print("Proxy有但Sersic NaN:", (df['Ellipticity Proxy'].notna() & df['Sersic Index (fitted)'].isna()).sum())

```
